```python
import sys
import matplotlib.pyplot as plt

# 1. Add 'src' folder to Python path so we can import our script cleanly
sys.path.append("../src")
from satellite_api import download_sentinel_bands, process_farm_indices

# 2. Define location and process all 3 indices
lat, lon = 36.7783, -119.4179
download_sentinel_bands(lat, lon, output_dir="../data")
indices = process_farm_indices(lat, lon, data_dir="../data")

# 3. Create a 1x3 Side-by-Side Plot
fig, axes = plt.subplots(1, 3, figsize=(20, 6))

# Plot 1: NDVI (Plant Health)
im0 = axes[0].imshow(indices['ndvi'], cmap='RdYlGn', vmin=-0.2, vmax=1.0)
axes[0].set_title("1. Plant Health (NDVI)", fontsize=14, fontweight='bold')
fig.colorbar(im0, ax=axes[0], fraction=0.046, pad=0.04)

# Plot 2: SAVI (Soil-Corrected Health)
im1 = axes[1].imshow(indices['savi'], cmap='YlGn', vmin=-0.2, vmax=1.0)
axes[1].set_title("2. Bare-Soil Adjusted (SAVI)", fontsize=14, fontweight='bold')
fig.colorbar(im1, ax=axes[1], fraction=0.046, pad=0.04)

# Plot 3: NDWI (Leaf Water Content / Stress)
im2 = axes[2].imshow(indices['ndwi'], cmap='YlGnBu', vmin=-0.5, vmax=0.8)
axes[2].set_title("3. Leaf Moisture (NDWI)", fontsize=14, fontweight='bold')
fig.colorbar(im2, ax=axes[2], fraction=0.046, pad=0.04)

plt.tight_layout()
plt.show()
```
